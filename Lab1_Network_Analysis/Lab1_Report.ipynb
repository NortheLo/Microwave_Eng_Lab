#### Return loss calculation from reflection coefficient:

```python
import numpy as np

def db_to_lin(dB):
    return 10**(dB / 20)

def return_loss(Gamma_lin):
    return -20 * np.log10(Gamma_lin)

# Measured reflection coefficient in dB
Gamma_dB = -5.968 

Gamma_lin = db_to_lin(Gamma_dB)
RL_dB = return_loss(Gamma_lin)

print(f"Return Loss = {RL_dB} dB")
```

## Task 3: Circulator

| Parameter     |   Measured Value     |
|---------------|--------|

| Isolation     |     $ -18 dB$   |

**Insertion Loss:** 
The amount of signal power lost in the DUT, expressed in dB.

$IL_{dB} \,  = 10 \cdot \log_{10}\left(\frac{P_{\text{in}}}{P_{\text{out}}}\right) = 20 \cdot \log_{10}\left(\frac{V_{\text{in}}}{V_{\text{out}}}\right)$

Since input and output matching is good ($S_{11}, S_{22}$ > 19dB), we can assume: 

$IL_{dB} \, = 20 \cdot \log_{10}\left(\frac{V_{\text{in}}}{V_{\text{out}}}\right) \approx 20 \cdot \log_{10}\left(\frac{1}{|S_{21,lin}|}\right)= -20 \cdot \log_{10}\left(|S_{21,lin}|\right)$ 

Or just $IL_{dB} \, = -|S_{21,dB}|$

**Return Loss:** 
Quantifies how much power is reflected back due to impedance mismatches in a transmission line or device.

$RL_{dB} \,  = 10 \cdot \log_{10}\left(\frac{P_{\text{incident}}}{P_{\text{reflected}}}\right) = 20 \cdot \log_{10}\left(\frac{V_{\text{incident}}}{V_{\text{reflected}}}\right) = -20 \cdot \log_{10}\left(|S_{11, lin}|\right) = -|S_{11,dB}|$

### 5.3 Measure the decoupling of the two outputs. Why does this change when you terminate the ports with $50\Omega$? Take a picture and send it at the end.

The quadrature hybrid coupler is commonly used for mismatch condition applications. An interesting function of the device is, that you can insert a signal at the input port, which splits ideally 50:50 to the output ports with 90 degree phase shift, if they are matched. If they are not matched, the reflected wave traveling back is going to the isolated port and is then, if the port is terminated, absorbed. 

This behaviour also works in the opposite direction. If 2 signals with 90 degree phase shift are injected at the output ports, they will combine to one inphase signal at the input port. If the input port is not matched, the reflected wave will travel to the isolated port and if terminated, the wave will be absorbed.

In figure 18 we can see, that the injected signal at one of the output ports is fully absorbed at the two terminated coupler ports. This is because the device works reciprocal and the injection at the output ports divides the signal 50:50 to the two terminated ports, which then fully absorbe.
For figure 17, the ports are not terminated. There, the injected signal travels 50:50 to the input port and isolated port and gets fully reflected with 90 degree phase shift. The reflected signals have half power each and are again split 50:50 by traveling to the two output ports of the coupler, where the measurement takes place. Therfore the isolation is reduced to theoretically 6dB but measurement shows around 7.25 dB.
