#### Return loss calculation from reflection coefficient

```python
import numpy as np

def db_to_lin(dB):
    return 10**(dB / 20)

def return_loss(Gamma_lin):
    return -20 * np.log10(Gamma_lin)

# Measured reflection coefficient in dB
Gamma_dB = -5.968 

Gamma_lin = db_to_lin(Gamma_dB)
RL_dB = return_loss(Gamma_lin)

print(f"Return Loss = {RL_dB} dB")
```
